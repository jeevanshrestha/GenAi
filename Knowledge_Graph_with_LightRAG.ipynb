### Data Preparation

```python

directory = os.getcwd()
file_directory = os.path.join(directory, 'Knowledge Graphs with LightRAG')
os.listdir(file_directory)
output_directory = os.path.join(file_directory, 'output')
os.makedirs(output_directory, exist_ok=True)
```

```python
# Open the pdf and extrat the text
pdf_text = []
file_path = os.path.join(file_directory,'Greek-gods-story.pdf')
with open(file_path, 'rb') as file:
    pdf_reader = PyPDF2.PdfReader(file)
    num_pages = len(pdf_reader.pages)
    for page in pdf_reader.pages:
        text = page.extract_text()
        pdf_text.append(text)
```

```python
# Save a file with the data
with open(f'{file_directory}/output/greek_gods.txt', 'w') as file:
    for item in pdf_text:
        file.write(f"{item}\n")
```

```python
# Build the RAG
with open(f'{file_directory}/output/greek_gods.txt', 'r') as file:
  rag.insert(file.read())
```

## RAG

```python
# Perform Naive RAG

rag = LightRAG(
    working_dir=WORKING_DIR,
    llm_model_func= gpt_4o_complete
)

def do_rag(query, mode):
  response = rag.query(query = query,
          param= QueryParam(mode = mode ))
  return Markdown(response)
```

```python
query = "who is Hermes?"
mode ="naive"
do_rag(query,mode)
```

### Perform Light RAG with a local search


```python
## Look into closer entities

mode = 'local'
do_rag(query, mode)
```

### Perform Light RAG with a Global search


```python
# More diverse results, global entity relationships

mode ='global'
do_rag(query, mode)
```

### Light RAG wit Hybrid Approch

```python
mode = 'hybrid'
do_rag(query, mode)
```

## Graph Visualization

```python
# Import libraries
import networkx as nx
from pyvis.network import Network
import matplotlib.pyplot as plt
import random
from IPython.display import HTML
```

```python
# Load Entities

G = nx.read_graphml(f"{WORKING_DIR}/graph_chunk_entity_relation.graphml")



# Create a Network
net = Network(notebook = True, cdn_resources = "remote")
net.from_nx(G)

# Add color and title to nodes
for node in net.nodes:
  node["color"] = "#{:06x}".format(random.randint(0, 0xFFFFFF))
  if "description" in node:
    node["title"] = node["description"]

# Add title to edges

for edge in  net.edges:
  if "description" in edge:
    edge['title'] = edge['description']



# Save and display the network
net.show("graph.html")
net.save_graph("graph.html")

HTML(filename='graph.html')

```
