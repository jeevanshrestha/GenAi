```python

import os
from openai import OpenAI
from IPython.display import Audio, display
import requests
```

```python
# Set your OpenAI API key here (or use Colab secrets)
OPENAI_API_KEY = os.getenv("genai_course")
```

```python

client = OpenAI(api_key=OPENAI_API_KEY)
```

```python


def text_to_speech(text, filename="output.mp3", voice="alloy", model="tts-1"):
    response = client.audio.speech.create(
        model=model,
        voice=voice,
        input=text
    )
    audio_url = response.url
    print(f"Audio URL: {audio_url}")

    # Download audio
    audio_data = requests.get(audio_url).content
    with open(filename, "wb") as f:
        f.write(audio_data)

    # Play audio in Colab
    display(Audio(filename, autoplay=True))

```

```python
from IPython.display import Audio, display



def text_to_speech(text, filename="output.mp3", voice="alloy", model="tts-1"):
    # Get the binary response object
    response = client.audio.speech.create(
        model=model,
        voice=voice,
        input=text
    )

    # Read the bytes from the response content
    audio_bytes = response.read()

    # Write to file
    with open(filename, "wb") as f:
        f.write(audio_bytes)

    print(f"Audio saved to {filename}")

    display(Audio(filename, autoplay=True))

# Example usage
text = "Hello, this is an example of text to speech using OpenAI's TTS API in Google Colab."
text_to_speech(text)
```
